```python
import torch
from torch import nn
```

```python
class SimpleMLP(nn.Module):
  def __init__(self,input_size,num_classes):
    super().__init__()
    self.ln1 = nn.Linear(in_features=input_size,out_features=25) # Weight Matrix-->(25,in_size) i.e 25 neurons
    self.relu = nn.ReLU()
    self.ln2 = nn.Linear(in_features=25,out_features=15) # Weight Matrix-->(15,25) i.e 15 neurons
    self.out = nn.Linear(in_features=15,out_features=num_classes) # Weight Matrix-->(15,25) i.e num_classes neurons
  def forward(self,x):
    x = self.ln1(x)
    x = self.relu(x)
    x = self.ln2(x)
    x = self.relu(x)
    out = self.out(x)
    return out
```

## Total number of parameters: \begin{equation} I*25+25+25*15+15+15*C+15 \end{equation}
where I = input_size,
C = num_classes

```python
model = SimpleMLP(input_size=100,num_classes=3)
sum(p.numel() for p in model.parameters())
```
